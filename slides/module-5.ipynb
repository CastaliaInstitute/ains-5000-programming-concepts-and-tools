# Worked example

```python
def estimated_cost(attendees, cost_per_person):
    return attendees * cost_per_person

attendance_counts = [12, 18, 24]
capacity = 20
for attendees in attendance_counts:
    if attendees <= capacity:
        room_status = "Room fits"
    else:
        room_status = "Choose a larger room"
    print(attendees, room_status, estimated_cost(attendees, 7.5))
```
